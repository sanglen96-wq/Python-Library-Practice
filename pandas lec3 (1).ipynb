```python

import pandas as pd

data = pd.DataFrame({
    'ID': [101, 102, 103, 104, 105],
    'NAME': ['RAM', 'SHYAM', 'RAHUL', 'AMIT', 'NEHA'],
    'SALARY': [35000, None, 30000, 40000, 50000]
})

data
```

```python
# we have start the replace method
print(data)
```

```python
data.replace({35000:30000,40000:50000})
```

```python
# target only column salary

#1. What does each part mean?
#- data → Your DataFrame name.
#- ['SALARY'] → Select the SALARY column.
#- .replace() → Replace existing values.
#- {30000:35000} → Replace 30000 with 35000.
#- {50000:40000} → Replace 50000 with 40000.

    
data['SALARY']=data['SALARY'].replace({30000:35000,50000:40000})
data
```

```python
import pandas as pd

data = pd.DataFrame({
    'ID': [101, 102, 103, 104, 105],
    'NAME': ['RAM', 'SHYAM', 'RAHUL', 'AMIT', 'NEHA'],
    'SALARY': [35000, None, 30000, 40000, 50000]
})

data
```

```python
print(data)
```

```python
# display name of the whose salary < name

data.loc[data['SALARY']>30000,'SALARY']

#explanation:
1)data['SALARY'] →    Selects the salary column.
2)> 35000 →           Checks salaries greater than 35,000.
3)'SALARY' →           Displays only the salary values.
```

```python
# if you > salary whose then you can apply the queries

data['SALARY'].max()
```

```python
#minimum salary

data['SALARY'].min()
```

```python
data[(data['ID'] > 22) & (data['SALARY'] > 50000)]

#Explanation
1. data['AGE'] > 22 → Checks age greater than 22.
2. data['SALARY'] > 50000 → Checks salary greater than 50,000.
3. & → Both conditions must be TRUE.
4. data[...] → Displays matching rows.
```
